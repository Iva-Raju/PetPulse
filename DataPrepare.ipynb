```python
import os
import shutil

# =========================
# PATH CONFIGURATION
# =========================
SOURCE_DATASET = "CAT_DB"           # main dataset folder
TARGET_DATASET = "animal_dataset"    # output folder

# =========================
# CLASS MAPPING
# =========================
STRESS_MAP = {
    "Angry": "Stressed",
    "Defense": "Stressed",
    "Fighting": "Stressed",
    "Paining": "Stressed",
    "Warning": "Stressed",
    "Happy": "Normal",
    "Resting": "Normal",
    "MotherCall": "Normal"
}

SKIP_CLASSES = {"HuntingMind", "Mating"}

# =========================
# CREATE OUTPUT FOLDERS
# =========================
for label in ["Stressed", "Normal"]:
    os.makedirs(os.path.join(TARGET_DATASET, label), exist_ok=True)

# =========================
# PROCESS DATASET
# =========================
total_copied = 0
total_skipped = 0

for class_name in os.listdir(SOURCE_DATASET):
    class_path = os.path.join(SOURCE_DATASET, class_name)

    if not os.path.isdir(class_path):
        continue

    # Skip unwanted classes
    if class_name in SKIP_CLASSES:
        print(f"SKIPPED class: {class_name}")
        total_skipped += len(os.listdir(class_path))
        continue

    # Map to binary label
    binary_label = STRESS_MAP.get(class_name)

    if binary_label is None:
        print(f"UNKNOWN class (ignored): {class_name}")
        continue

    target_class_path = os.path.join(TARGET_DATASET, binary_label)

    for file in os.listdir(class_path):
        if not file.lower().endswith((".wav", ".mp3", ".flac")):
            continue

        src_file = os.path.join(class_path, file)

        # Avoid filename collision
        new_name = f"{class_name}_{file}"
        dst_file = os.path.join(target_class_path, new_name)

        shutil.copy2(src_file, dst_file)
        total_copied += 1

print("\n=== DATASET CONVERSION COMPLETE ===")
print(f"Files copied   : {total_copied}")
print(f"Files skipped  : {total_skipped}")
print(f"Output folder  : {TARGET_DATASET}")
```

```python
import os
import shutil

# =========================
# PATHS
# =========================
SOURCE_DATASET = "Dog"
TARGET_DATASET = "animal_dataset"

# =========================
# DOG VOCALIZATION MAPPING
# =========================
STRESS_MAP = {
    "dog_growl": "Stressed",
    "dog_grunt": "Normal",
    "dog_bark": "Normal"
}

# =========================
# CREATE OUTPUT DIRS
# =========================
for label in ["Stressed", "Normal"]:
    os.makedirs(os.path.join(TARGET_DATASET, label), exist_ok=True)

# =========================
# PROCESS
# =========================
copied = 0

for folder in os.listdir(SOURCE_DATASET):
    folder_path = os.path.join(SOURCE_DATASET, folder)

    if not os.path.isdir(folder_path):
        continue

    # Identify base class (remove _train / _test)
    base_class = folder.replace("_train", "").replace("_test", "")

    binary_label = STRESS_MAP.get(base_class)

    if binary_label is None:
        print(f"SKIPPED folder: {folder}")
        continue

    target_path = os.path.join(TARGET_DATASET, binary_label)

    for file in os.listdir(folder_path):
        if not file.lower().endswith((".wav", ".mp3", ".flac")):
            continue

        src = os.path.join(folder_path, file)
        new_name = f"{base_class}_{file}"
        dst = os.path.join(target_path, new_name)

        shutil.copy2(src, dst)
        copied += 1

print("\n=== DOG DATASET CONVERSION DONE ===")
print(f"Files copied: {copied}")
print(f"Output path: {TARGET_DATASET}")
```
